```python
contacts = []

def ajouter_contact():
    nom = input("Nom : ")
    telephone = input("Téléphone : ")
    email = input("Email : ")

    # Vérifier si le nom existe déjà (éviter les doublons)
    for contact in contacts:
        if contact["nom"].lower() == nom.lower():
            print("Ce contact existe déjà !")
            return  
   #si le nom n'exxiste pas on creer un nouveau dicctionnaire
    nouveau_contact = {
        "nom": nom,
        "telephone": telephone,
        "email": email
    }

    contacts.append(nouveau_contact)
    print(f"Contact '{nom}' ajouté avec succès !")
```

```python
ajouter_contact()
ajouter_contact()
print(contacts)
```

```python
def afficher_contacts():
    if len(contacts) == 0:
        print("Aucun contact enregistré.")
        return

    print("--- Liste des contacts ---")
    for i, contact in enumerate(contacts, start=1):
        print(f"{i}. {contact['nom']} | {contact['telephone']} | {contact['email']}")
```

```python
afficher_contacts()
```

```python
def rechercher_contact():
    nom_recherche = input("Nom à rechercher : ")
    trouve = False

    for contact in contacts:
        if nom_recherche.lower() in contact["nom"].lower():
            print(f"{contact['nom']} | {contact['telephone']} | {contact['email']}")
            trouve = True

    if not trouve:
        print("Aucun contact trouvé.")
```

```python
rechercher_contact()
```

```python
def supprimer_contact():
    nom_a_supprimer = input("Nom du contact à supprimer : ")

    for contact in contacts:
        if contact["nom"].lower() == nom_a_supprimer.lower():
            contacts.remove(contact)
            print(f"Contact '{contact['nom']}' supprimé avec succès !")
            return

    print("Contact introuvable.")
```

```python
supprimer_contact()
```

```python
afficher_contacts()
```

```python
def modifier_contact()
    nom_a_modifier = input("Nom de contact a modifier:")
    
    for contact in contacts:
        if contact["nom"].lower() == nom_a_modifier.lower():
            print(f"contact trouve!{contact[nom]}|{contact[telephone]}|{contact[email]")

            nouveau_telephone =input("Entrez le nouveau numero de telephone (laissez vide si vous ne voulez plus modifier)")
            nouveau_email = input("Entrer le nouveau email (laissez vide si vous ne voulez plus modifier)")

            if nouveau_telephone !=""
              contact["telephone"] = nouveau_telephone
            if  nouveau_email !=""
              contact["email"] = nouveau_email
            print("Contact modifié avec succès !")
            return

    print("Contact introuvable.")
           

    
```
