```python
import pandas as pd

df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")
```

```python
print(df["Profit"])
```

```python
ventes_a_perte = df[df["Profit"] < 0]
print(ventes_a_perte.shape)
```

```python
print(df["Category"].unique())
```

```python
ventes_par_categorie = df.groupby("Category")["Sales"].sum()
print(ventes_par_categorie)
```

```python
top5 = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(5)
print(top5)
```

```python
print(df.isna().sum())
```

```python
managers = pd.DataFrame({
    "Region": ["West", "East", "Central", "South"],
    "Manager": ["Sadie", "Chuck", "Kelly", "Cassandra"]
})
print(managers)
```

```python
# fusion 
df_avec_manager = df.merge(managers, on="Region")
print(df_avec_manager.shape)
print(df_avec_manager[["Region", "Manager", "Sales"]].head())
```

```python
ventes_region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(ventes_region)

top10_produits = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
print(top10_produits)

profit_segment = df.groupby("Segment")["Profit"].sum().sort_values(ascending=False)
print(profit_segment)
```

```python
# on creer de feuilles de calcul differentes contenents  les informations sur les ventes par region ,le top 10 des produits,profit par segment
with pd.ExcelWriter("rapport_ventes.xlsx") as writer:
    ventes_region.to_excel(writer, sheet_name="Ventes par region")
    top10_produits.to_excel(writer, sheet_name="Top 10 produits")
    profit_segment.to_excel(writer, sheet_name="Profit par segment")

print("Rapport Excel généré !")
```

```python
verification = pd.read_excel("rapport_ventes.xlsx", sheet_name="Top 10 produits")
print(verification)
```

```python
print("Nombre de lignes (ventes de produits) :", df.shape[0])
print("Nombre de commandes uniques :", df["Order ID"].nunique())
print("Nombre de clients uniques :", df["Customer Name"].nunique())
```

```python
# analyse de la rentabiliter par segment
clients_par_segment = df.groupby("Segment")["Customer Name"].nunique()
print(clients_par_segment)

profit_par_client = profit_segment / clients_par_segment
print(profit_par_client.sort_values(ascending=False))
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Mois"] = df["Order Date"].dt.month

ventes_mois = df.groupby("Mois")["Sales"].sum().sort_values(ascending=False)
print(ventes_mois)

commandes_mois = df.groupby("Mois")["Order ID"].nunique().sort_values(ascending=False)
print(commandes_mois)
```
